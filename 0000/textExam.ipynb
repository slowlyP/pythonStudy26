```python
# 파일로 더미데이터 및 추가 자료를 관리해본다.
# 지금까지는 변수 선언할때 초기값을 리스트에 담아 제공 받음
# 그러나 프로그램을 종료하면 추가한 값은 사라진다.
# 자료를 영구적으로 보관하는 방법은 파일로 저장하던 데이터베이스에 보관을 한다.

# 서버란  365일 24시간 언제나 멈춤없이 서비스를 제공하는 컴퓨터
# 서비스 =  웹서비스,게임서비스,쇼핑서비스,등등
# 클라이언트 =  제공을 받는자

# 파일저장 :  텍스트 파일(메모장), csv(엑셀) , json(웹qpi)  등으로 저장이 가능하다.

```

```python
f = open("새파일.txt", "w")
# open = 파일을 열때 사용함.
# 옵션 = r -> read 읽기전용
#       w -> writer 쓰기가능
#       a -> append 추가가능 (마지막행에 추가)
# 파일을 열때 경로를 지정하지않으면 같은 위치에 파일을 찾거나 쓴다.
f.close()
# 파일은 윈도우 정책에서 단일처리를 한다. (열고(open) 작업이 끝나면 닫고(close) 나와야함
# 안닫으면 다른 프로그램에서 열수가없다.
```

```python
f = open("c:/mbc/새파일.txt","w")
f.close()
# 주의사항 =  경로에 / 를 붙이는게 일반적
# 만약 \를 쓰고싶으면 두번 써야함 c:\\mbc\\새파일.txt 이런식으로
# \t는 탭 \n는 엔터 기능 \는 이스케이프 문자 기능이있는 문자열
```

```python
f = open("c:/mbc/새파일.txt","w")
for i in range(1, 11) : # range (1에서 10까지 값을 가지고있음)
    data = "%d번째 줄입니다. \n" %i
    #       정수처리        엔터  for에 있는 i값
    f.write(data) # .write = 기록해라 라는 뜻
    # 파일에 저장한다 라는 뜻
f.close()
```

```python
# 파일에 있는 내용을 가져와보자.
# .readline() 파일에 있는 내용 줄단위로 가져오기
f = open("c:/mbc/새파일.txt","r") # r은 읽기전용
line = f.readline() # 기본값이 첫줄 가져오기임
print(line)
f.close()
```

```python
# 모든 줄을 읽어 화면에 출력하고싶으면 while 문 쓰면된다고함
f = open("c:/mbc/새파일.txt","r")
while True : # False 처리 될때 까지 반복문
    line = f.readline() # 한줄씩 읽는다
    if not line: break # line 가 없으면 break 중지 처리
    print(line)
f.close()
# 메모장내용에 엔터가 적용이 되어 있는 상태임 /n/r 라인피드 캐리지 리턴
```

```python
#.readlines() 메모장 파일을 리스트로 한번에 불러온다.
# 파일에 모든 줄을 읽어 리스트 형태로 반환함
# 반환값은 문자열 리스트 ['첫번째줄\n','두번째줄\n'....] 이런식으로 나옴
# 파일 내용을 모두 메모리에 로드하기 떄문에 파일크기가 클경우 메모리 부족현상이 나타날수있음
# 장점  =  파일의 모든 줄을 한번에 가져와 처리해야 할 때 편리하다고함
f = open("c:/mbc/새파일.txt","r")
lines = f.readlines() # 파일에 있는 모든 내용을 리스트 처리함
print(lines)
for line in lines : print(line)
f.close()
```

```python
# 메모장에 엔터효과인 \n 을 지우고 싶으면 strip 을 사용하면됨.
# strip은 문자열 내에서 원하는 문자열 또는 공백을 제거하는 역활
# lstrip() 원하는 문자열 또는 공백을 왼쪽에서 제거
# rstrip() 원하는 문자열 또는 공백을 오른쪽에서 제거

f = open("c:/mbc/새파일.txt","r")
lines = f.readlines()
for line in lines :
    line = line.strip() #  줄끝에 줄 바꿈 문자를 제거
    print(line)
f.close()
```

```python
# read() 를 이용해서 파일 문자열 읽기
# 파일의 내용 전체를 문자열로 반환
f = open("c:/mbc/새파일.txt","r")
data = f.read()
print(data) # for 문 사용하지않고 전체를 읽어 출력했음
f.close()
```

```python
# for 를 이용해 줄단위로 가져오기.
f = open("c:/mbc/새파일.txt","r")
# 파일 객체는 기본적으로 for 문과 같이 함께 사용하여 줄단위로 읽기도 가능함
for line in f:print(line)
f.close()
```

```python
# 파일에 새로운 내용을 추가 하려면 append -> a
# w로 파일을 열면 이미 존재하는 파일이면 그 내용이 사라지고 덮어쓰기가 된다고함.
# a 모드는 원래 있던 값을 유지하면서 새로운 값을 맨 아래에 추가 하는 역활
f = open("c:/mbc/새파일.txt","a") #append (추가모드)
f.write("------------------------")
for i in range(11,20):
    data = "%d번째 줄입니다. \n" %i
    f.write(data)
f.close()


```
