```python
# 딕셔너리 타입은 key : value 로 이루어진다.
# { key1 : value1, key2 : value2 ... } 이런식으로 씀
# 대부분 웹 프로그램 언어는 json 이라는 값으로 처리하는데
# 유사한 형태이다.

dic = {"name" : "smg" , 'age' : 34}
#       key     value    key   value 이렇게 된다.

type(dic)
print(dic["name"])
#           key
# 딕셔너리는 인덱스 대신에 key 를 가져온다.
```

```python
a = {1 : "Hi" , 2 : "Bye"}
print(a[1])
```

```python
b = {"a" : [1,2,3]} #key 에 리스트[]괄호를 넣어서 사용가능
#     key   value
print(b["a"])
print(b["a"][1])
```

```python
print(dic)
```

```python
# dic 변수에 email key를 갖는 값을 추가 하는 방법
dic["email"] = "qwe@ewq.com"
print(dic)
```

```python
# dic 변수에 age를 삭제방법
del dic['age']
print(dic)
```

```python
dic['name'] = "송명근"
print(dic)
```

```python
# key의 리스트를 생성하려면 :a.keys()
dic.keys()
```

```python
# key의 값들을 리스트로 생성하려면
dic.values()
```

```python
# key와 value의 쌍으로 튜플 처리하려면 items 사용하면됨
dic.items()
```

```python
# get 을 이용해 key 를 넣으면 값이 나오도록 하는것
dic.get("name")
```

```python
# key에 해당하는 값을 가져올때 없는 키를 넣으면 에러
dic["age"] # 없는 key를 가져오려면 오류가 발생한다.
# 이것은 프로그램 실행중 치명적인 오류로 강제 종료 된다.
```

```python
dic.get("age","값없음")
#        key    key가없을때의 출력문이 됨
```

```python
print(dic)
```

```python
"name" in dic # "in" 명령어는 리스트나, 튜플 딕셔너리 안에 값이 있는지 물어보는것
```

```python
"송명근" in dic # key만본다 value 는 안봄
```

```python
# key를 이용해서 value를 가져오는 다른방법 value - pop(값을 꺼내와서 변수에 넣고 삭제한다)
print(dic)
```

```python
email = dic.pop("email")
print(email)
```
