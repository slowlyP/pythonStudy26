```python
# 집합 자료형 (set)
# 로또와 같이 중복을 허용하지않고 순서가 없는
# 데이터들의 모임이다
# 교집합, 합집합, 차집합, 등의 집합 연산을 쉽게 처리 가능
```

```python
s1 = set([1, 2, 3, 4, 5, 6])
s2 = set([4, 5, 6, 7, 8, 9])
s3 = set("hello") #  중복된 ll은 하나만 출력되고 hello 는 순서가 랜덤으로 나온다.
print(s1)
print(s3)
```
