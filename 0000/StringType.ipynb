```python
# 주민 번호 앞 두자리 생년으로 추출 year
# 주민 번호 3~4번째 자리를 생월로 추출 month
# 주민 번호 5~6 자리를 생일로 추출 day
# 주민 번호 8째 한자리를 남여 추출 sex
```

```python

year = inputSsn[0:2]
month = inputSsn[2:4]
day = inputSsn[4:6]
sex = inputSsn[7:8]
print("귀하의 생년은 " + year + "년")
print("귀하의 생월은 " + month + "월")
print("귀하의 생일은 " + day + "일 입니다.")

# 주민번호를 이용해 남여 구분을 위한 코드
# 문자열로 인식된 숫자를 숫자로 계산하려면 타입변환 해야함
ssnNumber = inputSsn[7]
# print(ssnNumber)
# type(ssnNumber)
ssnNumber = int(ssnNumber) # 문자가 숫자로 타입변환된다. int
# type(ssnNumber)

if ssnNumber % 2 == 1 : # 숫자로된 성별이 2로 나눈 나머지가 1인지
    sex = "남자" + " 입니다."
else:
    sex = "여자" + " 입니다."
print(sex)

# 주민번호 8번째 글자가 1,2  5,  6이면 1900년생
# 3,4,6,7 이면 2000년생으로

if ssnNumber == 1 :
    yearif = "19"+year # 문자인경우 + 기호는 붙인다
elif ssnNumber == 2 :
    yearif = "19"+year
elif ssnNumber == 5 :
    yearif = "19"+year
elif ssnNumber == 6 :
    yearif = "19"+year
elif ssnNumber == 9 :
    print("존재하지않습니다.")

else:
    yearif = "20"+year

print(yearif)







```
