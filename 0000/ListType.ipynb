```python
# 리스트에 C R U D 기법을 적용해보기
# CREATE -> 생성,추가,삽입        .append()
# READ -> 전체 또는 한개 자료 보기  .index()
# UPDATE -> 자료 수정             passwords[3]<변경하려고하는곳 = "5555"하면됨
# DELETE -> 자료 삭제             del passwords[3]   or   .remove()    or   .pop()
```

```python
# 리스트에 자료를 추가해 보자
ids = ["smg", "lhj", "ksb"]
ids.append("MBC") #->맨뒤에 들어감   신규 가입시 .append(추가용) 사용하면될듯?
print(ids)
```

```python
passwords = ["1234","4321","4567"]
passwords.append("4444")
print(passwords)
```

```python
# 리스트에 있는 값 가져오기
ids.index("MBC")
```

```python
passwords[3]
# MBC id에 인덱스를 알아봐 패스워드값을 찾아온다
```

```python
# 알아온 pw 값을 5555로 수정해본다.
passwords[3] = "5555"
print(passwords)
```

```python
# 3번째 인덱스에 암호를 삭제 해보기.
passwords.remove("5555")
print(passwords)

```

```python
# pop은 삭제되는 내용을 출력하고 삭제한다.
# passwords.pop() 는 맨 마지막 자료 삭제
# passwords.pop(2) 는 2번 인덱스를 삭제한다
passwords.pop(2)
```
